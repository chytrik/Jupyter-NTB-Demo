# OpenShift AI MaaS Demo: Detekce objektů na obrázku

```python
import requests
import json
import urllib3
from IPython.display import Image, display

# Potlačení SSL varování pro self-signed certifikáty (odpovídá curl -k)
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
```

```python
IMAGE_URL = "https://ekonomickydenik.cz/app/uploads/2025/09/p1049810-3.jpg"

print("Předložený obrázek pro analýzu:")
display(Image(url=IMAGE_URL, width=600))
```

```python
import os
MODEL_URL = os.environ.get("MODEL_URL",
    "https://qwen25-vl-7b-kserve-workload-svc.rhoai-playground.svc.cluster.local:8000/v1")
ENDPOINT_URL = MODEL_URL + "/chat/completions"
MODEL_NAME = "RedHatAI/Qwen2.5-VL-7B-Instruct-quantized.w4a16"

headers = {
    "Content-Type": "application/json"
}

payload = {
    "model": MODEL_NAME,
    "messages": [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Kolik vojáků je na obrázku?"},
                {"type": "image_url", "image_url": {"url": IMAGE_URL}}
            ]
        }
    ],
    "max_tokens": 512
}
```

```python
response = requests.post(
    ENDPOINT_URL,
    headers=headers,
    json=payload,
    verify=False
)

if response.status_code == 200:
    result = response.json()
    answer = result["choices"][0]["message"]["content"]
    print("--- ODPOVĚĎ MODELU ---")
    print(answer)
else:
    print(f"Chyba {response.status_code}:")
    print(response.text)
```
