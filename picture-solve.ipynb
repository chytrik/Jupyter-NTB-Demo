# OpenShift AI MaaS Demo: Detekce objektů na obrázku

```python
import requests
import json
import urllib3
from IPython.display import Image, display

# Potlačení SSL varování pro self-signed certifikáty (odpovídá curl -k)
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
```

```python
IMAGE_URL = "https://ekonomickydenik.cz/app/uploads/2025/09/p1049810-3.jpg"

print("Předložený obrázek pro analýzu:")
display(Image(url=IMAGE_URL, width=600))
```

```python
import os
ENDPOINT_URL = "https://maas.apps.ocp.b6ngg.sandbox1066.opentlc.com/muse-glimmer/muse-glimmer/v1/chat/completions"
# API klíč se načítá z prostředí, aby se nedostal do repozitáře:
#   export MAAS_API_KEY="sk-oai-..."
API_KEY = os.environ["MAAS_API_KEY"]
MODEL_NAME = "muse-glimmer"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

payload = {
    "model": MODEL_NAME,
    "messages": [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Kolik vojáků je na obrázku?"},
                {"type": "image_url", "image_url": {"url": IMAGE_URL}}
            ]
        }
    ],
    "max_tokens": 512
}
```

```python
response = requests.post(
    ENDPOINT_URL,
    headers=headers,
    json=payload,
    verify=False
)

if response.status_code == 200:
    result = response.json()
    answer = result["choices"][0]["message"]["content"]
    print("--- ODPOVĚĎ MODELU ---")
    print(answer)
else:
    print(f"Chyba {response.status_code}:")
    print(response.text)
```
